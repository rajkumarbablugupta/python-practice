```python
# sort list  Aliphonumureically
#list object have  a sort() METHOD THAT WILL SORT THE LIST ALPHANUMERICALLY ,ACENDING BY DEFAULT
```

```python
# ASC
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)
```

```python
##sort Decending
thislist=["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse=True)
print(thislist)
```

```python
thislist=["banana","orange","Kiwi","cherry"]
thislist.reverse()
print(thislist)
```

```python
#copy  a list
#You cannot  copy list simply by typing list2=list1,because:list2 will only be reference to
##There are way to make coppy ,one way is to use the built-in list method copy().
```

```python
## MAKE a copy of a list with  the copy()method:
thislist=["apple","banana","cherry"]
mylist=thislist.copy()
print(mylist)
```

```python
## makea copy of list with the list ()method:
thislist=["apple","banana","cherry"]
mylist=list(thislist)
print(mylist)
```

```python
##count    (count the no .of cherry)
fruits=["apple","banana","cherry"]
x=fruits.count("cherry")
print(x)
```

```python
##index What is the  position of hte values "cherry";
fruits=["apple","banana","cherry"]
x=fruits.index("cherry")
print(x)
```

```python
##Append list2 into list1;
list1=["a","b","c"]
list2=[1,2,3]

for x in list2:
    list1.append(x)
    print(list1)
```

```python
## Use the extent()method to add list 2 at the end of list1:
list1=["a","b","c"]
list2=[1,2,3]

list1.extend(list2)
print(list1)
```
