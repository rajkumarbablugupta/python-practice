```python
class car:
    def __init__(self,name,model):
        self.name=name
        self.model=model

    def move(self):
      print("run")

class boat:
    def __init__(self,name,model):
        self.name=name
        self.model=model

    def move(self):
        print("sail")

class plane:
    def __init__(self,name,model):
        self.name=name
        self.model=model


    def move(self):
        print("fly")

car1 = car("Ford", "Mustang")       #Create a Car class
boat1 = boat("Ibiza", "Touring 20") #Create a Boat class
plane1 = plane("Boeing", "747")     #Create a Plane class

for x in (car1, boat1, plane1):
    x.move()

print(x.move())

print(car1.model)
```

```python
##Encapsulation in Python



class base:
    def __init__(self):
        self.a="rajkumar"
        self.b="gupta"
class derived(base):
    def __init__(self):
        base.__init__(self)
        print("calling private member of base class")
        print(self.a)
obj1=base()
obj2=derived()
print(obj1.a)
print(obj2.b)
```

```python
from abc import ABC,abstractmethod
class car(ABC):
  def __init__(self,brand,model,year,speed):
    self.brand=brand
    self.model=model
    self.year=year
    self.speed=speed
    self.engine_status=False

  def start_engine(self):
    if not self.engine_status:
       print(f"starting the {self.model} engine")
       self.engine_status = True
    else:
       print("engine is already started")
c=car("ford","mustang",2023,200)
c.start_engine()
```

```python
# Import required modules
from abc import ABC, abstractmethod

# Create Abstract base class
class Car(ABC):
  def __init__(self, brand, model, year,age):
    self.brand = brand
    self.model = model
    self.year = year
    self.age = age

  # Create abstract method
  @abstractmethod
  def printDetails(self):
    pass

  # Create concrete method
  def accelerate(self):
    print("speed up ...")

  def break_applied(self):
    print("Car stop")

# Create a child class
class Hatchback(Car):

  def printDetails(self):
    print("Brand:", self.brand);
    print("Model:", self.model);
    print("Year:", self.year);
    print("age:", self.age);

  def Sunroof(self):
    print("Not having this feature")

# Create a child class
class Suv(Car):

  def printDetails(self):
    print("Brand:", self.brand);
    print("Model:", self.model);
    print("Year:", self.year);
    print("age:", self.age)

  def Sunroof(self):
    print("Available")


car1 = Hatchback("Maruti", "Alto", "2022","45");

car1.printDetails()
car1.break_applied()
```
